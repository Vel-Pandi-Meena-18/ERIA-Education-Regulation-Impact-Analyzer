````python
import fitz
import pytesseract
from PIL import Image
import io

pdf_path = "/content/UGC Reg.pdf"

doc = fitz.open(pdf_path)

english_text = ""

for page_number in range(8, 14):
    page = doc[page_number]

    pix = page.get_pixmap(matrix=fitz.Matrix(2, 2))
    image = Image.open(io.BytesIO(pix.tobytes("png")))

    page_text = pytesseract.image_to_string(image, lang="eng")
    english_text += page_text + "\n"

print("Total characters extracted:", len(english_text))
print("\n--- EXTRACTED TEXT PREVIEW ---\n")
print(english_text[:5000])
````

````python
from pypdf import PdfReader

pdf_path = "/content/Internship-Research-Guidelines.pdf"

reader2 = PdfReader(pdf_path)

print("Number of pages:", len(reader2.pages))

text2 = ""

for page in reader2.pages:
    page_text = page.extract_text()
    if page_text:
        text2 += page_text + "\n"

print("Total characters extracted:", len(text2))
print("\n--- PREVIEW ---\n")
print(text2[:3000])
````

## Text Processing


````python
import re

def clean_text(text):
    # Remove page-break characters
    text = text.replace("\x0c", "\n")

    # Replace multiple spaces/tabs with a single space
    text = re.sub(r"[ \t]+", " ", text)

    # Reduce excessive blank lines
    text = re.sub(r"\n\s*\n+", "\n\n", text)

    # Remove leading/trailing spaces from each line
    text = "\n".join(line.strip() for line in text.splitlines())

    return text.strip()


clean_regulation_text = clean_text(english_text)
clean_guideline_text = clean_text(text2)

print("Regulation characters:", len(clean_regulation_text))
print("Guideline characters:", len(clean_guideline_text))

print("\n--- CLEAN REGULATION PREVIEW ---\n")
print(clean_regulation_text[:3000])
````

````python
import os
import glob
import re
import pymupdf
import pytesseract
from PIL import Image
from pypdf import PdfReader

# Show PDFs currently uploaded
pdf_files = glob.glob("/content/*.pdf")

print("PDFs found:")
for f in pdf_files:
    print("-", os.path.basename(f))


# --------------------------------------------------
# 1. FIND THE UGC REGULATION PDF
# --------------------------------------------------

regulation_path = None

for f in pdf_files:
    if "UGC" in os.path.basename(f).upper() and "REG" in os.path.basename(f).upper():
        regulation_path = f
        break

if regulation_path is None:
    raise FileNotFoundError("UGC Regulation PDF was not found in /content.")

print("\nRegulation PDF:", os.path.basename(regulation_path))


# --------------------------------------------------
# 2. OCR THE REGULATION PDF
# --------------------------------------------------

doc = pymupdf.open(regulation_path)

english_text = ""

for page_number, page in enumerate(doc, start=1):

    pix = page.get_pixmap(matrix=pymupdf.Matrix(2, 2))

    img = Image.frombytes(
        "RGB",
        [pix.width, pix.height],
        pix.samples
    )

    page_text = pytesseract.image_to_string(
        img,
        lang="eng"
    )

    english_text += page_text + "\n"

doc.close()

print("Regulation OCR characters:", len(english_text))


# --------------------------------------------------
# 3. CLEAN REGULATION TEXT
# --------------------------------------------------

def clean_text(text):
    text = text.replace("\x0c", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n\n", text)
    text = "\n".join(line.strip() for line in text.splitlines())
    return text.strip()


clean_regulation_text = clean_text(english_text)

print("Clean regulation characters:", len(clean_regulation_text))


# --------------------------------------------------
# 4. FIND THE INTERNSHIP GUIDELINE PDF
# --------------------------------------------------

guideline_path = None

for f in pdf_files:
    name = os.path.basename(f).lower()

    if "internship" in name or "guideline" in name:
        guideline_path = f
        break

if guideline_path is not None:

    print("\nGuideline PDF:", os.path.basename(guideline_path))

    reader = PdfReader(guideline_path)

    text2 = ""

    for page in reader.pages:
        page_text = page.extract_text()

        if page_text:
            text2 += page_text + "\n"

    clean_guideline_text = clean_text(text2)

    print("Guideline characters:", len(clean_guideline_text))

else:
    print("\nGuideline PDF not found.")


# --------------------------------------------------
# 5. CREATE OUR STRUCTURED DOCUMENT
# --------------------------------------------------

regulation_document = {
    "document_name": "UGC Minimum Standards of Instruction Regulations, 2025",
    "document_type": "Regulation",
    "source": "University Grants Commission (UGC)",
    "text": clean_regulation_text
}


print("\n==============================")
print("REGULATION DOCUMENT READY")
print("==============================")
print("Name:", regulation_document["document_name"])
print("Type:", regulation_document["document_type"])
print("Source:", regulation_document["source"])
print("Text length:", len(regulation_document["text"]))
````
